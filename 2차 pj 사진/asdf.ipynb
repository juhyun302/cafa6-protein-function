```python
import os

# 파일 경로 설정 (같은 폴더에 있다고 가정)
INPUT_ALL = "fasta_accessions.txt"           # 전체 리스트
INPUT_EXCLUDE = "no_alphafold_structure.txt" # 뺄 놈들
OUTPUT_FILE = "fasta_accessions_filtered.txt" # 결과 파일

def filter_ids():
    print("🚀 필터링 작업을 시작합니다...")

    # 1. 제외할 ID들을 먼저 메모리에 로드 (Set 사용 -> 검색 속도 O(1)로 매우 빠름)
    exclude_set = set()
    with open(INPUT_EXCLUDE, "r", encoding="utf-8") as f:
        for line in f:
            pid = line.strip()
            if pid: # 빈 줄이 아니면 추가
                exclude_set.add(pid)
    
    print(f"📋 제외할 ID 개수(No Structure): {len(exclude_set)}개")

    # 2. 전체 파일을 읽으면서 필터링 후 저장
    kept_count = 0
    total_count = 0
    
    with open(INPUT_ALL, "r", encoding="utf-8") as f_in, \
         open(OUTPUT_FILE, "w", encoding="utf-8") as f_out:
        
        for line in f_in:
            pid = line.strip()
            if not pid: continue # 빈 줄 건너뛰기
            
            total_count += 1
            
            # 제외 리스트에 없으면(구조가 있으면) 저장
            if pid not in exclude_set:
                f_out.write(pid + "\n")
                kept_count += 1

    # 3. 결과 리포트
    print("-" * 30)
    print(f"📊 원본 ID 개수 : {total_count}")
    print(f"❌ 제외된 ID 개수: {total_count - kept_count}")
    print(f"✅ 남은 ID 개수 : {kept_count}")
    print(f"💾 저장 완료: {OUTPUT_FILE}")

if __name__ == "__main__":
    filter_ids()
```
