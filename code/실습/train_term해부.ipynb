# train_terms.tsv 해부

단백질 하나당 GO term 몇 개?  
희귀 GO가 얼마나 많은가?  
상위 1% GO가 라벨의 몇 %를 차지하는가?  
BP / MF / CC 난이도 차이

#### 위를 통해  
왜 asymmetric loss 이야기가 나오는지  
왜 threshold가 모델만큼 중요한지  
왜 CAFA는 accuracy로 평가 못 하는지

EntryID : 단백질 ID
term : GO term (예: GO:0005515)
aspect : ontology 구분
	P = Biological Process (BP)
	F = Molecular Function (MF)
	C = Cellular Component (CC)

#### 1) 전체 규모 (이게 “라벨 세계관” 크기)
전체 라벨 행(row) 수: 537,027

GO term 종류 수(고유 term 수): 26,125

#### 2) “단백질 1개당 라벨이 몇 개냐?” (희소성의 핵심)
평균: 6.52개
중앙값: 4개
p90: 14개

대부분 단백질은 라벨이 4개 정도로 “적은 편”
근데 일부는 수십~수백 개 라벨을 가짐 → 분포가 매우 비대칭
그래서 “그냥 정확도/Top-k” 같은 감각으로는 모델을 판단하기가 어렵고,
threshold/불균형 처리가 성능을 결정하는 구조야.

#### 4) 롱테일(불균형) 정도 — “이게 진짜 핵심”

📌 해석(진짜 중요):
GO term의 대부분이 매우 희귀하다
“상위 몇 개 쉬운 라벨”만 잘 맞추면 겉보기 성능은 나오지만,
CAFA 점수(F-max, IA 반영 등)에서는 희귀/정보량 높은 term을 어느 정도 건져야 점수가 올라감
그래서 너가 예전에 물어본 asymmetric loss / focal / class weight 같은 게 등장하는 거야
(데이터가 그렇게 강요함)
