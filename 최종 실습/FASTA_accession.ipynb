# FASTA에서 Accession 목록만 뽑기

1. 앞으로 모든 파일을 A0A0C5B5G6 처럼 일관된 이름으로 저장해야 함  
2. Foldseek/AlphaFold 다운로드/파일명 정리

```python
from pathlib import Path

fasta_path = Path(r"C:\Users\EL19\Downloads\cafa-6-protein-function-prediction\cafa-6-protein-function-prediction\Test\testsuperset.fasta")  # 너 경로로 바꿔
out_path = Path("fasta_accessions.txt")

accessions = []

with open(fasta_path, "r", encoding="utf-8") as f:
    for line in f:
        if line.startswith(">"):
            # 예: >sp|A0A0C5B5G6|MOTSC_HUMAN ...
            parts = line.strip().split("|")
            if len(parts) >= 3:
                acc = parts[1]
                accessions.append(acc)
            else:
                # 예외(형식 다른 헤더) 대비
                accessions.append(line[1:].split()[0])

# 중복 제거(혹시 모를 상황 대비)
accessions = sorted(set(accessions))

with open(out_path, "w", encoding="utf-8") as f:
    for acc in accessions:
        f.write(acc + "\n")

print("Accession 개수:", len(accessions))
print("샘플 5개:", accessions[:5])
print("저장 파일:", out_path.resolve())
```
